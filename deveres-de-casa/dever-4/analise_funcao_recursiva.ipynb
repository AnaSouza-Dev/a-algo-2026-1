Considando a seguinte função recursiva:

F(n)=2F(n−1)+n^2
com o caso base F(1)=2

Tarefa:
1. Implemente essa função em Python usando recursão.
2. O programa deve solicitar ao usuário um valor de n e calcular F(n).

Bibliotecas

```python
import math
```

Função recursiva

```python
def calcular_f(n):
    #caso base 
    if n == 1:
        return 2
    # chamda recursiva
    else:
        return 2 * calcular_f(n -1) + pow(n, 2)
```

Função principal para interação com o usuário.

```python
def main():
    """
    Função principal para interação com o usuário.
    """
    try:
        n = int(input("Digite o valor de n (inteiro positivo): "))

        if n < 1:
            print("Por favor, insira um número maior ou igual a 1.")
        else:
            resultado = calcular_f(n)
            print(f"O valor de F({n}) é: {resultado}")

    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro.")

        
if __name__ == "__main__":
    main()
```

# Complexidade de tempo: O algoritmo é O(n), pois realiza n chamadas recursivas, cada uma com tempo constante de processamento.
# Complexidade de Espaço: Também é O(n) devido à pilha de recursão (call stack). Para valores muito altos de n, pode atingir o RecursionError.
