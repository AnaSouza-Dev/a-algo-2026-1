1. Algoritmo de ordenação Merge Sort

```python
def merge_sort(arr):
    # Condição de parada: se a lista tiver 1 ou 0 elementos, já está ordenada
    if len(arr) > 1:
        # Encontra o meio da lista
        meio = len(arr) // 2
        
        # Divide a lista em duas metades
        metade_esquerda = arr[:meio]
        metade_direita = arr[meio:]

        # Chamada recursiva para ordenar as duas metades
        merge_sort(metade_esquerda)
        merge_sort(metade_direita)

        # Índices para percorrer as metades e a lista principal
        i = j = k = 0

        # Intercala as duas metades ordenadas
        while i < len(metade_esquerda) and j < len(metade_direita):
            if metade_esquerda[i] < metade_direita[j]:
                arr[k] = metade_esquerda[i]
                i += 1
            else:
                arr[k] = metade_direita[j]
                j += 1
            k += 1

        # Verifica se sobrou algum elemento na metade esquerda
        while i < len(metade_esquerda):
            arr[k] = metade_esquerda[i]
            i += 1
            k += 1

        # Verifica se sobrou algum elemento na metade direita
        while j < len(metade_direita):
            arr[k] = metade_direita[j]
            j += 1
            k += 1
            
    return arr

# Exemplo de uso
lista = [38, 27, 43, 3, 9, 82, 10]
print("Lista original:", lista)
print("Lista ordenada:", merge_sort(lista))
```

Merge Sort — T(n) = Θ(n log n)

A recorrência do Merge Sort é T(n) = 2T(n/2) + n.

A divisão: em cada chamada, o vetor é partido ao meio (custo O(1)), e a fusão (merge) percorre os dois subvetores, custando Θ(n).

A árvore tem log₂ n níveis (porque dividimos por 2 a cada passo).
Em cada nível, o trabalho total de fusão é sempre cn (a soma de todos os subproblemas naquele nível é proporcional a n).

Somando todos os níveis: n × log₂ n = Θ(n log n).

2. Multiplicação de Matrizes

```python
def multiplicar_matrizes(A, B):
    # Assume que A e B são matrizes quadradas n x n
    n = len(A)
    
    # Cria uma matriz resultante C de tamanho n x n preenchida com zeros
    C = [[0 for _ in range(n)] for _ in range(n)]

    # Percorre as linhas da matriz A
    for i in range(n):
        # Percorre as colunas da matriz B
        for j in range(n):
            # Calcula o produto escalar da linha i de A pela coluna j de B
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]
                
    return C

# Exemplo de uso
matriz_A = [
    [1, 2],
    [3, 4]
]

matriz_B = [
    [5, 6],
    [7, 8]
]

resultado = multiplicar_matrizes(matriz_A, matriz_B)

print("Resultado da Multiplicação:")
for linha in resultado:
    print(linha)
```

A multiplicação tradicional de duas matrizes quadradas de tamanho n X  n gera uma nova matriz de tamanho n x n.

Análise: O algoritmo exige calcular os produtos e somas para cada um dos n^2 elementos da nova matriz.

Para calcular um único elemento, é necessário iterar n vezes (multiplicando a linha de uma matriz pela coluna da outra).
 
Loops: Isso resulta em 3 laços de repetição aninhados, cada um executando n vezes (n x n x n = n^3).

Complexidade Final: O(n^3)

3. Solucionador de Recorrências (Teorema Mestre)

```python
def resolver_teorema_mestre(a, b, exp_f):
    """
    Resolve recorrências no formato T(n) = a*T(n/b) + O(n^exp_f)
    """
    # Calcula log_b(a)
    log_b_a = math.log(a, b)
    
    # Tolerância para lidar com imprecisões de ponto flutuante em Python
    tol = 1e-9 
    
    # Compara o expoente de f(n) com log_b(a)
    if abs(exp_f - log_b_a) < tol:
        # Caso 2: f(n) tem o mesmo grau que n^(log_b_a)
        if exp_f == 0.5:
            return "Theta(√n * log n)"
        elif exp_f == 1:
            return "Theta(n * log n)"
        else:
            return f"Theta(n^{exp_f} * log n)"
            
    elif exp_f > log_b_a:
        # Caso 3: f(n) cresce mais rápido
        if exp_f == 1:
            return "Theta(n)"
        else:
            return f"Theta(n^{exp_f})"
            
    else:
        # Caso 1: n^(log_b_a) cresce mais rápido
        if log_b_a == 0.5:
            return "Theta(√n)"
        else:
            return f"Theta(n^{log_b_a})"

# Testando os casos do seu exercício:
print("Equação A: T(n) = 2T(n/4) + √n")
print("Complexidade:", resolver_teorema_mestre(a=2, b=4, exp_f=0.5))
print("-" * 30)

print("Equação B: T(n) = 2T(n/4) + n")
print("Complexidade:", resolver_teorema_mestre(a=2, b=4, exp_f=1.0))
print("-" * 30)

print("Equação C: T(n) = 16T(n/4) + n²")
print("Complexidade:", resolver_teorema_mestre(a=16, b=4, exp_f=2.0))
```
