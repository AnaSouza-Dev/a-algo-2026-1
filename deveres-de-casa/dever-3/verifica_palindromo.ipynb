# Esse notebook Implementa o código recursivo para verificar se um array é palíndromo.

Bibliotecas

```python
import time
import matplotlib.pyplot as plt
```

Função Recursiva

```python
def eh_palindromo(arr, esquerda, direita):
    # Caso Base: se os ponteiros se cruzarem, é palíndromo
    if esquerda >= direita:
        return True
    
    # Se os elementos forem diferentes, não é palíndromo
    if arr[esquerda] != arr[direita]:
        return False
    
    # Chamada recursiva para os elementos internos
    return eh_palindromo(arr, esquerda + 1, direita -1)
```

Definição dos testes

```python
testes = {
    "array1": [0, 1, 2, 3, 2, 1, 0],
    "array2": ["a", "b", "b", "a"],
    "array3": ["a", "b", "c", "b", "a"],
    "array4": ["a", "b", "c", "f", "b", "a"]
}

```

Configuração para medição

```python

NUMERO_REPETICOES = 1000
nomes_grafico = []
tempos_medios_grafico = []
```

```python

print(f"{'Teste':<10} | {'Resultado':<18} | {'Tempo Gasto (segundos)':<25}")
print("-" * 60)

for nome, lista in testes.items():
    inicio = time.perf_counter()
    
    # Executa o teste 1000 vezes para obter uma média estável
    for _ in range(NUMERO_REPETICOES):
        eh_palindromo(lista, 0, len(lista) - 1)
        
    fim = time.perf_counter()
    
    # Calcula a média por execução
    tempo_medio = (fim - inicio) / NUMERO_REPETICOES
    
    nomes_grafico.append(nome)
    tempos_medios_grafico.append(tempo_medio)
    
    print(f"{nome:<10} | {len(lista):<8} | {tempo_medio:.10f}")
```

Geração do Gráfico

```python
plt.figure(figsize=(8, 5))
plt.bar(nomes_grafico, tempos_medios_grafico, color=['blue', 'green', 'orange', 'red'])
plt.ylabel('Tempo Médio em Segundos')
plt.title(f'Tempo Médio de Execução (Média de {NUMERO_REPETICOES} rodadas)')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

# O  gasto de memória cresce linearmente com o tamanho do array: O(n)
